```python
import numpy as np
#2x4 array
a = np.array([
    [1 ,3 ,4 ,5 ,6 ],
    [2, 3, 4, 5, 6],
    [1, 4, 5, 6 ,7]
              ])

b = np.array([
    [[1,2,3,4,],
     [2,4,5,6]],
    [[1,3,6,9],
     [4,7,9,0]]
])
print(a)
print(a.ndim)
print(a.shape)
print(a[0,1:6])
print(a[0,:])
print(a[1,0])
print(a[1,2])
print(a[1,4])
print(a[1])
print(a[:,2])
print(a[:,1])
print(a[2:,])
print(a[:, 1])
print(b[[0, 1],[1, 0]])

#reshaping
arr = np.array([1,3,4,5,6,6,7,7,7,9,8 ])
print(arr.ndim)
print(arr.shape)
print(arr.reshape(11,1))
print(arr.reshape(1,11))
print(arr.reshape(-1,1))
print(b.reshape(-1,1))
print(arr.size)
print(b.size)

#boolean
ages = np.array([12,28 ,25,30, 15])
ages >=18
ages [ages >=18]

patient = np.array([[30,25,40],
                   [45,18,55,]])
print(patient >= 45)
print(patient[patient >=45])
patient[:, 0]
```
